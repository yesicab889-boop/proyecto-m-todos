```python
lugares = [
    "Quebrada La Honda",
    "Quebrada El Hato",
    "Rio Bogota",
    "Rio Sumapaz",
    "Rio Frio",
    "Rio Apulo",
    "Quebrada La Campana",
    "Quebrada El Tigre",
    "Rio Ubate",
    "Quebrada Socota"
]

ph = [4.6, 5.5, 7.0, 8.5, 7.6, 3.9, 4.8, 6.7, 6.1, 3.5]
```

```python
for i in range(len(ph)):
    if ph[i] < 5:
        print(lugares[i], ph[i])
```

```python
import pandas as pd
muestra = pd.DataFrame({"lugares":["Rio Bogota"], "ph":[7.0]})
valo_ph = muestra.loc[0,"ph"]
if valor_ph <6.5:
    clasificación = "acido"
elif valor_ph <= 8.5:
    clasificación = "neutro"
else:
    clasificación = "basico"
print(f"lugares {muestra.loc[0, "lugares"]}: agua {clasificación}")
```

```python
muestra_lista = ["Quebrada Socota", 3.5]
valor_ph = muestra_lista[1]
if valor_ph <= 4.0:
  alerta = ("Riesgo de metales pesados")
else:
  alerta = "Niveles estables de acidez."
print(f"Lugar: {muestra_lista[0]} (pH: {valor_ph}) = {alerta}")
```

```python
muestra_tupla = ("Rio Frio", 7.6)
valor_ph = muestra_tupla[1]
if 6.5 <= valor_ph <= 8.4:
  riego = "Apto para uso agrícola y riego"
else:
  riego = "Requiere tratamiento antes de ser usado en cultivos"
print(f"Lugar: {muestra_tupla[0]} (pH: {valor_ph}) = {riego}")
```

```python
muestra_dict = {"lugar": "Quebrada La Honda", "ph": 4.6}
valor_ph = muestra_dict["ph"]
limite_norma = 8.0
if valor_ph > limite_norma:
  cumplimiento = ("Incumple norma ambiental: pH demasiado alcalino por vertimientos")
else:
  cumplimiento = "Cumple con los parámetros permitidos"
print(f"Lugar: {muestra_dict['lugar']} (pH: {valor_ph}) = {cumplimiento}")
```

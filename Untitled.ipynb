Ejercicio 3: Dos nombres para los mismos datos
Primero en papel:

revision = original no crea una nueva lista, sino una segunda etiqueta que apunta al mismo lugar de memoria.   

Al modificar revision[0] = 7.0, también cambia original[0].   

Ambas salidas mostrarán: [7.0, 7.2, 6.5].   

Cambio para imaginar: Al usar revision = original.copy(), se crea una copia independiente. Ahí sí se conserva el registro original intacto ([6.8, 7.2, 6.5])


```python
# Versión 1: Asignación por referencia
original = [6.8, 7.2, 6.5]
revision = original
revision[0] = 7.0
print("Original sin copy:", original)
print("Revisión sin copy:", revision)

# Versión 2: Copia independiente
original_2 = [6.8, 7.2, 6.5]
revision_2 = original_2.copy()
revision_2[0] = 7.0
print("Original con .copy():", original_2)
print("Revisión con .copy():", revision_2)
```

Ejercicio 4: Qué operación realiza el signo más
Primero en papel:

El signo + concatena (une) las dos listas: datos = [20.0, 22.0, 18.0, 20.0].   

Suma total = 80.0. Cantidad = 4.

Salida de la media: 20.0.   

Cambio para imaginar: Si se usa un divisor fijo como / 2 o / 4, la media fallará al cambiar el número de muestras. La expresión sum(datos)/len(datos) adapta el divisor dinámicamente

Ejercicio 5: Cambiar una coordenada
Primero en papel:

punto = (4.6097, -74.0817) crea una tupla (inmutable).   

print(punto[0]) mostrará 4.6097.   

La línea punto[1] = -74.0820 provocará un error de tipo TypeError porque las tuplas no permiten modificar sus elementos internos.

```python
punto = (4.6097, -74.0817)
print(punto[0])

# Intentar modificar un elemento individual de la tupla causa un error
try:
    punto[1] = -74.0820
except TypeError as e:
    print("Error detectado:", e)

# Para cambiar la ubicación conservando una tupla, se debe reasignar la variable completa:
punto = (4.6097, -74.0820)
print("Tupla reasignada:", punto)
```

Ejercicio 9: Listas paralelas o registros completos
Primero en papel:

Salida: M001 6.5.   

Problema ambiental: El pH 6.5 le pertenecía a M002. Al ordenar la lista ph sola, se desvinculó de su ID, cometiendo un error grave de trazabilidad de datos.

```python
  # Programa original con error metodológico
ids = ["M001", "M002", "M003"]
ph = [7.2, 6.5, 8.0]
ph.sort()
print("Incorrecto (Listas desvinculadas):", ids[0], ph[0])

# Solución correcta: Usar lista de diccionarios
muestras = [
    {"id": "M001", "ph": 7.2},
    {"id": "M002", "ph": 6.5},
    {"id": "M003", "ph": 8.0}
]
muestras_ordenadas = sorted(muestras, key=lambda x: x["ph"])
print("Correcto (Registro completo):", muestras_ordenadas[0])
```

Ejercicio 10: De una muestra a una lista de muestras

Ejercicio 11: De registros a una tabla

Ejercicio 13: Qué cambia al cambiar la condición
Primero en papel:

df["ph"] > 7.5 sólo selecciona la muestra M002 (7.8). M001 tiene exactamente 7.5 y es excluida por ser una condición estricta.   
+ 1

Cambiar a >= incluiría a M001 y M002.

```python
df = pd.DataFrame({
    "id": ["M001", "M002", "M003"],
    "ph": [7.5, 7.8, 6.9],
    "temperatura_c": [20.0, 22.0, 21.0]
})

print("Filtro estricto (> 7.5):")
print(df.loc[df["ph"] > 7.5, ["id", "ph"]])

print("\nFiltro inclusive (>= 7.5):")
print(df.loc[df["ph"] >= 7.5, ["id", "ph"]]) 
```

Ejercicio 14: Ordenar y conservar el resultado
Primero en papel:

La línea df.sort_values(...) devuelve una vista ordenada pero no modifica la variable df.   

df["id"].tolist() imprimirá el orden original: ['M001', 'M002', 'M003'].   

Para guardar el nuevo orden hay que reasignar: df = df.sort_values(...).

```python
df = pd.DataFrame({
    "id": ["M001", "M002", "M003"],
    "temperatura_c": [20.5, 18.3, 22.0]
})

# Sin reasignar:
df.sort_values("temperatura_c")
print("Sin reasignar:", df["id"].tolist())

# Reasignando la variable:
df_ordenado = df.sort_values("temperatura_c")
print("Reasignado correctamente:", df_ordenado["id"].tolist())
```

Ejercicio 15: Una columna calculada y los paréntesis

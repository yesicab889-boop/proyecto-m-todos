```python
import pandas as pd

# 1. Tabla con lugares y temperaturas inventadas (°C)
muestra = pd.DataFrame({
    "lugares": ["Río Bogotá", "Río Magdalena", "Laguna de Tota", "Río Amazonas"],
    "temperatura": [14.5, 28.0, 9.5, 26.5]
})

# 2. Función y condicionales para clasificar la temperatura
def clasificar_temperatura(temp):
    if temp < 12.0:
        return "fría"
    elif temp <= 22.0:
        return "templada"
    else:
        return "cálida"

# 3. Bucle para recorrer cada fila y mostrar el resultado
for i in range(len(muestra)):
    lugar = muestra.loc[i, "lugares"]
    valor_temp = muestra.loc[i, "temperatura"]
    estado = clasificar_temperatura(valor_temp)
    
    print(f"Lugar: {lugar} | Temperatura: {valor_temp} °C -> Agua {estado}")

```
